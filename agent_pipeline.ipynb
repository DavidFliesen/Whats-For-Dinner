# What's 4 Dinner? Autonomous AI Research Agent
This Google Colab notebook uses Gemini 2.5 Flash to automatically discover local restaurants, scrape/verify nutrition disclosures, filter for GLP-1 suitability (<500mg sodium, lean digestible protein), and export directly to `meals.json`.

```python
!pip install -q google-genai pydantic

import os
import json
from google import genai
from pydantic import BaseModel, Field

# Insert your Gemini API key
os.environ['GEMINI_API_KEY'] = 'YOUR_API_KEY_HERE'
client = genai.Client()
```

```python
TARGET_LOCATION = 'San Francisco, CA'  # Adjust to your neighborhood or city

prompt = f'''Act as a clinical sports dietitian specializing in GLP-1 receptor agonist patients.
Analyze dining spots in {TARGET_LOCATION} with meals meeting:
1. Under 500mg sodium (strict DASH).
2. Over 30g lean protein.
3. Free of severe nausea triggers (no heavy fried oils, minimal sulfurous bloat).
Provide 5 options formatted strictly as a JSON array matching the What's 4 Dinner schema.'''

response = client.models.generate_content(
    model='gemini-2.5-flash',
    contents=prompt,
)

with open('meals.json', 'w') as f:
    f.write(response.text)

print('Wrote curated nutritional data to meals.json!')
```
